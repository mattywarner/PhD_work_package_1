```python
import numpy as np
```

```python
tau = np.array([[[np.ones(12)*(-5),np.ones(12)*6,np.ones(12)*9],
                [np.ones(12)*2,np.ones(12)*(-3),np.ones(12)*4],
                [np.ones(12)*7,np.ones(12)*8,np.ones(12)*(-10)]],
                [[np.ones(12)*(-5),np.ones(12)*6,np.ones(12)*9],
                [np.ones(12)*2,np.ones(12)*(-3),np.ones(12)*4],
                [np.ones(12)*7,np.ones(12)*8,np.ones(12)*(-10)]],
                [[np.ones(12)*(-5),np.ones(12)*6,np.ones(12)*9],
                [np.ones(12)*2,np.ones(12)*(-3),np.ones(12)*4],
                [np.ones(12)*7,np.ones(12)*8,np.ones(12)*(-10)]],
                ])

gamma = np.array([[[np.ones(12)*5,np.ones(12)*6,np.ones(12)*9],
                  [np.ones(12)*2,np.ones(12)*3,np.ones(12)*4],
                  [np.ones(12)*7,np.ones(12)*8,np.ones(12)*10]],
                  [[np.ones(12)*5,np.ones(12)*6,np.ones(12)*9],
                  [np.ones(12)*2,np.ones(12)*3,np.ones(12)*4],
                  [np.ones(12)*7,np.ones(12)*8,np.ones(12)*10]],
                  [[np.ones(12)*5,np.ones(12)*6,np.ones(12)*9],
                  [np.ones(12)*2,np.ones(12)*3,np.ones(12)*4],
                  [np.ones(12)*7,np.ones(12)*8,np.ones(12)*10]]
                  ])
```

```python
tau.shape, gamma.shape
```

```python
plastic_work = np.abs(np.multiply(tau,gamma))
```

```python
accumulated_plastic_work = np.sum(plastic_work, axis = 3)
accumulated_plastic_work.shape
```

```python
accumulated_plastic_work
```

```python
full_arr = np.array([accumulated_plastic_work]*12)
```

```python
full_arr.shape
```

```python
full_arr_sum = np.sum(full_arr, axis=0)
full_arr_sum.shape, full_arr_sum
```
