```python
import CPanalysistools
from CPanalysistools import *
```

```python
pheno_alpha_vol_frac = np.loadtxt("./phenopowerlaw/rolled_copper_polycrystal_200_grains_alpha_fibre_vol_frac_pheno.txt")
pheno_beta_vol_frac = np.loadtxt("./phenopowerlaw/rolled_copper_polycrystal_200_grains_beta_fibre_vol_frac_pheno.txt")
pheno_texture_indices = np.loadtxt("./phenopowerlaw/rolled_copper_polycrystal_200_grains_texture_indices_pheno.txt")
```

```python
dislo_alpha_vol_frac = np.loadtxt("./dislo_density/rolled_copper_polycrystal_200_grains_alpha_fibre_vol_frac_dislo.txt")
dislo_beta_vol_frac = np.loadtxt("./dislo_density/rolled_copper_polycrystal_200_grains_beta_fibre_vol_frac_dislo.txt")
dislo_texture_indices = np.loadtxt("./dislo_density/rolled_copper_polycrystal_200_grains_texture_indices_dislo.txt")
```

```python
pheno_equiv_vM_strains = np.load("../data/vM_strains/pheno_vM_strains_avg_20_time_steps.npy")
dislo_equiv_vM_strains = np.load("../data/vM_strains/dislo_vM_strains_avg_20_time_steps.npy")
```

```python
%matplotlib inline
fig_comp, ax_comp = plt.subplots(1,2, figsize = (20,8))

ax_comp[0].plot(pheno_equiv_vM_strains, pheno_alpha_vol_frac*100, 'mediumorchid', marker = 'x', label = 'Pheno alpha fibre')
ax_comp[0].plot(pheno_equiv_vM_strains, pheno_beta_vol_frac*100, 'mediumorchid', marker = '>', label = 'Pheno beta fibre')
ax_comp[0].plot(dislo_equiv_vM_strains, dislo_alpha_vol_frac*100, 'green', marker = 'x', label = 'Dislo alpha fibre')
ax_comp[0].plot(dislo_equiv_vM_strains, dislo_beta_vol_frac*100, 'green', marker = '>', label = 'Dislo beta fibre')

ax_comp[1].plot(pheno_equiv_vM_strains, pheno_texture_indices, 'mediumorchid', marker = 'o', label = 'Phenopowerlaw')
ax_comp[1].plot(dislo_equiv_vM_strains, dislo_texture_indices, 'green', marker = 'o', label = 'Dislocation density')

ax_comp[0].legend(shadow = True, fontsize = 12)
ax_comp[1].legend(shadow = True, fontsize = 12)
```
